```python
data_with_missing = {
    "cars": ["Ferrari", "BMD", None, "Tesla", None, "BWD"],
    "counts": [2, 4, None, 7, 4, None]
}

df_missing = pd.DataFrame(data_with_missing)
print(df_missing)
```

```python
df_missing.isnull()
```

```python
df_missing.isnull().sum()
# gives how many nonvaluables in each column
```

```python
df_missing.dropna()
# deletes all = missing values
```

```python
df_missing.dropna(how="all")
# first colum has nunevaluable? => delete
```

```python
df_missing.dropna(axis=1)
```

```python
df_missing.dropna(subset=["counts"])
# count qiymatiga asoslanadi
```

```python
df_missing.dropna(subset=["cars"])
# cars qiymatiga asoslanadi
```

```python
df_missing.fillna(0)
# missing qismga belgilangan narsani qoyib toldiradi
```

```python
df_missing.fillna("unknown")
```

```python
df_missing.fillna("<-->")
```

```python
df_missing.fillna({"cars":"unknown", "counts": "0"})
```
