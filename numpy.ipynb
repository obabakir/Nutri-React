```python
import numpy as np

arr = np.array([1, 3, 5, 7])
print('numpy_array:', arr)

arr.shape
```

```python
arr2 = np.array([[1, 3, 5, 7], [4, 6, 8, 9]])
print('numpy_array:', arr2)

arr2.shape
```

```python
# dtype()
a = np.array([1, 2, 3])
print(a.dtype)
```

```python
# reshape()
a = np.array([1, 2, 3, 4, 5, 6, ])
b = a.reshape(2, 3)
print('b:', b)
```

```python
# astype()
a = np.array([1, 2, 3])
b = a.astype(float)
print('float:', b)
```

```python
# np.zeros(), np.ones() => dummu data
z = np.zeros((2,3))
o = np.ones((2,3))
print('zeros:', z)
print('ones:', o)
```

```python
# ndim()
a =np.array([[1, 2, 3, 4], [5, 6, 7, 8],  [3, 4, 5, 6]])
print('ndim:', a.ndim)
print(a.shape)
```

```python
# arange() dan => gacha
a = np.arange(1, 7)
print('arange:', a)
```

```python
# sum() => yigindisi, mean() => ortachasi, max() => maximum, min() => minimum
a = np.array([1, 2, 3, 4])
print('sum:', a.sum())
print('mean:', a.mean())
print('max:', a.max())
print('min:', a.min())
```

```python
# flatten() 2d, 3d ni => 1d ga ogiradi
a =np.array([[1, 2], [3, 4]])
print('flatten:', a.flatten())
```

```python
# T(Transpose) arraylar elementlarini joylashuvini ozgartirib beradi
a = np.array([[1, 2], [3, 4]])
print('transpose:', a.T)
```

```python
# dot() @
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print('dot, multiplication:', a.dot(b))
```

```python
# random.rand()
a = np.random.rand(2, 3)
print('random_rand:', a)

# from 0.0 =>  to 1.0/this is excluded
```
